PHẦN 4: THỐNG KÊ SUY DIỄN (3 ĐIỂM)




1.	Thực hiện kiểm định trung bình của 1 biến số (định lượng) bằng phương pháp T-Test của 1 một thuộc tính nào đó với mức sai lầm là 5%. Cho nhận xét


```python
import scipy.stats as stats

alpha = 0.05  # Mức sai lầm 5%
mu = 70  # Giá trị trung bình cần kiểm định

# T-Test
t_stat, p_value = stats.ttest_1samp(df['Homework 2'], mu)

print(f"T-Statistic: {t_stat}, P-Value: {p_value}")

# Nhận xét
if p_value < alpha:
    print("Có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình khác với", mu)
else:
    print("Không có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình bằng", mu)
```

2.	Thực hiện kiểm định trung bình của 2 biến số (định lượng) có bằng nhau không bằng phương pháp Z-Test với mức sai lầm là 10%. Cho nhận xét


```python
# Giả sử df chứa hai cột cần kiểm định 'Homework 1' và 'Homework 2'
alpha = 0.10  # Mức sai lầm 10%

# Tính toán trung bình và độ lệch chuẩn
mean1 = df['Homework 1'].mean()
mean2 = df['Homework 2'].mean()
std1 = df['Homework 1'].std(ddof=1)
std2 = df['Homework 2'].std(ddof=1)
n1 = len(df['Homework 1'])
n2 = len(df['Homework 2'])

# Z-Test
z_stat = (mean1 - mean2) / ((std1**2 / n1) + (std2**2 / n2))**0.5
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print(f"Z-Statistic: {z_stat}, P-Value: {p_value}")

# Nhận xét
if p_value < alpha:
    print("Có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của hai biến không bằng nhau")
else:
    print("Không có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của hai biến bằng nhau")
```

3.	Thực hiện kiểm tra 2 biến định lượng có tương quan với nhau không bằng phương pháp Fisher với mức sai lầm là 5%? Cho nhận xét.


```python
# Giả sử df có hai cột 'Homework 3' và 'Homework 2'
alpha = 0.05  # Mức sai lầm 5%

# Tính toán hệ số tương quan
corr_coef, p_value = stats.pearsonr(df['Homework 3'], df['Homework 2'])

print(f"Hệ số tương quan: {corr_coef}, P-Value: {p_value}")

# Nhận xét
if p_value < alpha:
    print("Có đủ chứng cứ để bác bỏ giả thuyết H0: Hai biến không tương quan")
else:
    print("Không có đủ chứng cứ để bác bỏ giả thuyết H0: Hai biến có tương quan")
```

4.	Thực hiện kiểm định ANOVA 1 chiều và 2 chiều. 

```python
df['Group'] = ['High' if score >= 70 else 'Low' for score in df['Homework 2']]
```

```python
import scipy.stats as stats

# Nhóm dữ liệu theo cột 'Group'
group_high = df[df['Group'] == 'High']['Homework 2']
group_low = df[df['Group'] == 'Low']['Homework 2']

# Kiểm định ANOVA 1 chiều
f_stat, p_value = stats.f_oneway(group_high, group_low)

print(f"F-Statistic: {f_stat}, P-Value: {p_value}")

# Nhận xét
alpha = 0.05  # Mức sai lầm 5%
if p_value < alpha:
    print("Có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của các nhóm khác nhau")
else:
    print("Không có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của các nhóm bằng nhau")
```

```python
# Tạo hai cột phân loại giả định
df['Gender'] = ['Male' if i % 2 == 0 else 'Female' for i in range(len(df))]
df['Factor'] = ['A' if i < 10 else 'B' for i in range(len(df))]
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Thực hiện ANOVA 2 chiều
model = ols('`Homework 1` ~ C(Gender) + C(Factor) + C(Gender):C(Factor)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)

# Nhận xét
alpha = 0.05  # Mức sai lầm 5%
for factor in anova_table.index:
    if anova_table.loc[factor, 'PR(>F)'] < alpha:
        print(f"Có đủ chứng cứ để bác bỏ giả thuyết H0 đối với {factor}")
    else:
        print(f"Không có đủ chứng cứ để bác bỏ giả thuyết H0 đối với {factor}")
```
