PHẦN 4: THỐNG KÊ SUY DIỄN (3 ĐIỂM)

```python
# 1.	Thực hiện kiểm định trung bình của 1 biến số (định lượng) bằng phương pháp T-Test của 1 một thuộc tính nào đó với mức sai lầm là 5%. Cho nhận xét
from scipy.stats import ttest_1samp
alpha = 0.05
# Kiểm định giả thuyết
# H0: Giá trị trung bình của Exam 1 là 50
# H1: Giá trị trung bình của Exam 1 không phải là 50
t_statistic, p_value = ttest_1samp(df['Exam 1'], 50)
print(f'T-Statistic: {t_statistic}')
print(f'P-Value: {p_value}')
# Nhận xét:
if p_value < alpha:
    print('Có bằng chứng để bác bỏ giả thuyết H0')
else:
    print('Chưa có bằng chứng để bác bỏ giả thuyết H0')

```

```python
# 2.	Thực hiện kiểm định trung bình của 2 biến số (định lượng) có bằng nhau không bằng phương pháp Z-Test với mức sai lầm là 10%. Cho nhận xét
from statsmodels.stats.weightstats import ztest
alpha = 0.1

# Kiểm định giả thuyết
# H0: Giá trị trung bình của Exam 1 bằng với giá trị trung bình của Exam 2
# H1: Giá trị trung bình của Exam 1 không bằng với giá trị trung bình của Exam 2
z_statistic, p_value = ztest(df['Exam 1'], df['Exam 2'])
print(f'Z-Statistic: {z_statistic}')
print(f'P-Value: {p_value}')
# Nhận xét
if p_value < alpha:
    print("Có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của hai biến không bằng nhau")
else:
    print("Không có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của hai biến bằng nhau")

```

```python
# 3.	Thực hiện kiểm tra 2 biến định lượng có tương quan với nhau không bằng phương pháp Fisher với mức sai lầm là 5%? Cho nhận xét.
from scipy.stats import pearsonr

# Kiểm định giả thuyết
# H0: Exam 1 và Exam 2 không tương quan với nhau
# H1: Exam 1 và Exam 2 tương quan với nhau
alpha = 0.05
correlation, p_value = pearsonr(df['Exam 1'], df['Exam 2'])
print(f'Correlation: {correlation}')
print(f'P-Value: {p_value}')
# Nhận xét
if p_value < alpha:
    print("Có đủ chứng cứ để bác bỏ giả thuyết H0: Hai biến không tương quan")
else:
    print("Không có đủ chứng cứ để bác bỏ giả thuyết H0: Hai biến có tương quan")
```

```python
# 4.	Thực hiện kiểm định ANOVA 1 chiều và 2 chiều. 
import scipy.stats as stats

# Nhóm dữ liệu theo cột 'Group'
group_high = df[df['SID'] == 'High']['Exam 2']
group_low = df[df['SID'] == 'Low']['Exam 2']

# Kiểm định ANOVA 1 chiều
f_stat, p_value = stats.f_oneway(group_high, group_low)

print(f"F-Statistic: {f_stat}, P-Value: {p_value}")

# Nhận xét
alpha = 0.05  # Mức sai lầm 5%
if p_value < alpha:
    print("Có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của các nhóm khác nhau")
else:
    print("Không có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của các nhóm bằng nhau")





```

```python
# Tạo hai cột phân loại giả định
df['Group'] = np.where(df['Exam 1'] > 50, 'High', 'Low')
df['Group'] = np.where(df['Exam 2'] > 50, 'High', 'Low')
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Rename the columns to remove spaces
df.rename(columns={'Exam 1': 'Exam1', 'First Name': 'FirstName'}, inplace=True)

# Ensure 'Group' and 'FirstName' are categorical variables
df['Group'] = df['Group'].astype('category')
df['FirstName'] = df['FirstName'].astype('category')

# Check for missing values
print(df.isnull().sum())

# Drop rows with missing values in 'Exam1', 'Group', or 'FirstName'
df = df.dropna(subset=['Exam1', 'Group', 'FirstName'])

# Verify the data
print(df.head())

# Thực hiện ANOVA 2 chiều
try:
    model = ols('Exam1 ~ C(Group) + C(FirstName) + C(Group):C(FirstName)', data=df).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)
    print(anova_table)

    # Nhận xét
    alpha = 0.05  # Mức sai lầm 5%
    if anova_table['PR(>F)'][0] < alpha:
        print("Có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của các nhóm khác nhau")
    else:
        print("Không có đủ chứng cứ để bác bỏ giả thuyết H0: Trung bình của các nhóm bằng nhau")
except Exception as e:
    print(f"An error occurred: {e}")
```
