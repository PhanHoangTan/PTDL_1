```python
# 2. Thực hiện kiểm định trung bình của 2 biến số (định lượng) có bằng nhau không bằng phương pháp T-Test với mức sai lầm là 5%. Cho nhận xét
from scipy.stats import ttest_ind
# Mức sai lầm (alpha)
alpha = 0.05

# Kiểm định giả thuyết
# Giả thuyết H0: Trung bình của biến số 'CRIM' và 'ZN' bằng nhau
# Giả thuyết H1: Trung bình của biến số 'CRIM' và 'ZN' không bằng nhau

# Thực hiện kiểm định T-Test
t_statistic, p_value = ttest_ind(df['CRIM'], df['ZN'])

# Kết quả
print(f'T-Statistic: {t_statistic}')
print(f'P-Value: {p_value}')

# Nhận xét chi tiết
if p_value < alpha:
    print("Bác bỏ giả thuyết không (H0). Trung bình của hai biến số khác nhau.")
    print(f"Giá trị p-value là {p_value:.4f}, nhỏ hơn mức ý nghĩa {alpha}.")
    print("Điều này cho thấy có sự khác biệt đáng kể về trung bình giữa hai biến số 'CRIM' và 'ZN'.")
else:
    print("Không bác bỏ giả thuyết không (H0). Trung bình của hai biến số không khác nhau.")
    print(f"Giá trị p-value là {p_value:.4f}, lớn hơn mức ý nghĩa {alpha}.")
    print("Điều này cho thấy không có sự khác biệt đáng kể về trung bình giữa hai biến số 'CRIM' và 'ZN'.")
```

```python
# 3. Thực hiện kiểm tra 2 biến định lượng có tương quan với nhau không bằng phương pháp Chi-Square với mức sai lầm là 10%? Cho nhận xét.
# Chuyển đổi các biến định lượng thành các biến định tính bằng cách phân loại chúng thành các nhóm (bins)
from scipy.stats import chi2_contingency
df['CRIM_binned'] = pd.cut(df['CRIM'], bins=3, labels=['Low', 'Medium', 'High'])
df['ZN_binned'] = pd.cut(df['ZN'], bins=3, labels=['Low', 'Medium', 'High'])

# Tạo bảng tần số chéo (contingency table)
contingency_table = pd.crosstab(df['CRIM_binned'], df['ZN_binned'])

# Thực hiện kiểm định Chi-Square
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Mức sai lầm (alpha)
alpha = 0.10

# Kết quả
print(f'Chi-Square Statistic: {chi2}')
print(f'P-Value: {p}')

# Nhận xét chi tiết
if p < alpha:
    print("Bác bỏ giả thuyết không (H0). Có sự tương quan giữa hai biến số.")
    print(f"Giá trị p-value là {p:.4f}, nhỏ hơn mức ý nghĩa {alpha}. Điều này cho thấy có sự tương quan đáng kể giữa 'CRIM' và 'ZN'.")
    print("Điều này có nghĩa là phân phối của 'CRIM' thay đổi theo các mức của 'ZN'.")
else:
    print("Không bác bỏ giả thuyết không (H0). Không có sự tương quan giữa hai biến số.")
    print(f"Giá trị p-value là {p:.4f}, lớn hơn mức ý nghĩa {alpha}. Điều này cho thấy không có sự tương quan đáng kể giữa 'CRIM' và 'ZN'.")
    print("Điều này có nghĩa là phân phối của 'CRIM' không thay đổi theo các mức của 'ZN'.")
```

```python
from scipy.stats import f_oneway
# Kiểm định ANOVA một chiều
# Giả sử bạn muốn kiểm định ANOVA cho biến 'CRIM' theo các nhóm của biến 'RAD'
# Tạo các nhóm dựa trên biến 'RAD'
group1 = df[df['RAD'] == 1]['CRIM']
group2 = df[df['RAD'] == 2]['CRIM']
group3 = df[df['RAD'] == 3]['CRIM']
group4 = df[df['RAD'] == 4]['CRIM']
group5 = df[df['RAD'] == 5]['CRIM']
group6 = df[df['RAD'] == 6]['CRIM']
group7 = df[df['RAD'] == 7]['CRIM']
# Thực hiện kiểm định ANOVA một chiều
f_statistic, p_value = f_oneway(group1, group2, group3, group4, group5, group6, group7)
# Mức sai lầm (alpha)
alpha = 0.05
# Kết quả
print(f'F-Statistic: {f_statistic}')
print(f'P-Value: {p_value}')
# Nhận xét chi tiết
if p_value < alpha:
    print("Bác bỏ giả thuyết không (H0). Có sự khác biệt đáng kể giữa các nhóm.")
    print(f"Giá trị p-value là {p_value:.4f}, nhỏ hơn mức ý nghĩa {alpha}.")
    print(f"Giá trị F-statistic là {f_statistic:.4f}, cho thấy có sự khác biệt đáng kể giữa các nhóm 'RAD' về giá trị 'CRIM'.")
else:
    print("Không bác bỏ giả thuyết không (H0). Không có sự khác biệt đáng kể giữa các nhóm.")
    print(f"Giá trị p-value là {p_value:.4f}, lớn hơn mức ý nghĩa {alpha}.")
    print(f"Giá trị F-statistic là {f_statistic:.4f}, cho thấy không có sự khác biệt đáng kể giữa các nhóm 'RAD' về giá trị 'CRIM'.")
```

```python
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

# Thực hiện kiểm định ANOVA hai chiều
# Giả sử bạn muốn kiểm định ảnh hưởng của 'RAD' và 'CHAS' lên 'CRIM'
model = ols('CRIM ~ C(RAD) + C(CHAS) + C(RAD):C(CHAS)', data=df).fit()
anova_results = anova_lm(model, typ=2)

# Kết quả
print(anova_results)

# Nhận xét chi tiết
alpha = 0.05
for factor in anova_results.index:
    p_value = anova_results.loc[factor, 'PR(>F)']
    f_statistic = anova_results.loc[factor, 'F']
    if p_value < alpha:
        print(f"Bác bỏ giả thuyết không (H0) cho yếu tố {factor}. Có sự khác biệt đáng kể.")
        print(f"Giá trị p-value cho yếu tố {factor} là {p_value:.4f}, nhỏ hơn mức ý nghĩa {alpha}.")
        print(f"Giá trị F-statistic cho yếu tố {factor} là {f_statistic:.4f}, cho thấy yếu tố này có ảnh hưởng đáng kể đến biến phụ thuộc 'CRIM'.")
    else:
        print(f"Không bác bỏ giả thuyết không (H0) cho yếu tố {factor}. Không có sự khác biệt đáng kể.")
        print(f"Giá trị p-value cho yếu tố {factor} là {p_value:.4f}, lớn hơn mức ý nghĩa {alpha}.")
        print(f"Giá trị F-statistic cho yếu tố {factor} là {f_statistic:.4f}, cho thấy yếu tố này không có ảnh hưởng đáng kể đến biến phụ thuộc 'CRIM'.")
```

```python
# 5. Xây dựng mô hình hồi quy tuyến tính dự báo 1 thuộc tính dựa trên biến độc lập và 1 biến phụ thuộc.
import statsmodels.api as sm

# Xây dựng mô hình hồi quy tuyến tính dự báo 'MEDV' dựa trên 'RM'
X = df['RM']  # Biến độc lập
y = df['MEDV']  # Biến phụ thuộc

# Thêm cột hằng số vào X để phù hợp với mô hình hồi quy tuyến tính
X = sm.add_constant(X)

# Xây dựng mô hình
model = sm.OLS(y, X).fit()

# Kết quả
print(model.summary())

# Viết phương trình hồi quy
print("Phương trình hồi quy:")
print(f"MEDV = {model.params[0]:.4f} + {model.params[1]:.4f} * RM")
```
