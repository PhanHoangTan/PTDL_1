```python
# 56. Tạo 1 file chương trình hiện các menu gồm các mục trên và mục cuối là thoát. Người 
# dùng chọn chức năng nào trong menu thì chương trình sẽ thực hiện chức năng tương 
# ứng.


import pandas as pd

# Sample DataFrame
df = pd.DataFrame({
    'SALES': [1000, 2000, 1500],
    'QUANTITYORDERED': [10, 20, 15],
    'PRICEEACH': [100, 200, 150],
    'ORDERLINENUMBER': [1, 2, 3],
    'YEAR_ID': [2023, 2022, 2021],
    'MONTH_ID': [10, 9, 8],
    'PRODUCTLINE': ['Classic Cars', 'Motorcycles', 'Planes'],
    'MSRP': [150, 250, 350],
    'PRODUCTCODE': ['S10_1678', 'S10_1949', 'S10_2016'],
    'CUSTOMERNAME': ['John Doe', 'Jane Smith', 'Alice Johnson'],
    'PHONE': ['123-456-7890', '987-654-3210', '555-555-5555'],
    'ADDRESSLINE1': ['123 Elm St', '456 Oak St', '789 Pine St'],
    'ADDRESSLINE2': ['Apt 456', '', 'Suite 101'],
    'CITY': ['Anytown', 'Othertown', 'Sometown'],
    'STATE': ['CA', 'TX', 'NY'],
    'POSTALCODE': ['90210', '73301', '10001'],
    'COUNTRY': ['USA', 'USA', 'USA'],
    'TERRITORY': ['NA', 'NA', 'NA'],
    'CONTACTLASTNAME': ['Doe', 'Smith', 'Johnson'],
    'CONTACTFIRSTNAME': ['John', 'Jane', 'Alice'],
    'DEALSIZE': ['Small', 'Medium', 'Large'],
    'FLAG': ['Y', 'N', 'Y'],
    'WORTH': [5000, 10000, 7500],
    'QTR_ID': [4, 3, 2],
    'YEAR_BIRTH': [1980, 1990, 1985]
})

# Function to describe a column
def describe_column(column_name):
    if column_name in df.columns:
        print(f"\nDescription of '{column_name}':\n", df[column_name].describe())
    else:
        print(f"Column '{column_name}' does not exist in the DataFrame")

# Function to add a new record
def add_record():
    new_record = {
        'SALES': 3000, 'QUANTITYORDERED': 30, 'PRICEEACH': 300, 'ORDERLINENUMBER': 4, 
        'YEAR_ID': 2024, 'MONTH_ID': 11, 'PRODUCTLINE': 'Trains', 'MSRP': 450, 
        'PRODUCTCODE': 'S10_2021', 'CUSTOMERNAME': 'Bob Brown', 'PHONE': '111-222-3333', 
        'ADDRESSLINE1': '101 Maple St', 'ADDRESSLINE2': '', 'CITY': 'Newcity', 
        'STATE': 'FL', 'POSTALCODE': '33101', 'COUNTRY': 'USA', 'TERRITORY': 'NA', 
        'CONTACTLASTNAME': 'Brown', 'CONTACTFIRSTNAME': 'Bob', 'DEALSIZE': 'Small', 
        'FLAG': 'N', 'WORTH': 6000, 'QTR_ID': 1, 'YEAR_BIRTH': 1975
    }
    global df
    df = df.append(new_record, ignore_index=True)
    print("New record added successfully!")

# Function to optimize memory usage
def optimize_memory():
    print("\nMemory usage before optimization:")
    df.info(memory_usage='deep')

    # Optimizing column types
    df['SALES'] = df['SALES'].astype('int32')
    df['QUANTITYORDERED'] = df['QUANTITYORDERED'].astype('int16')
    df['PRICEEACH'] = df['PRICEEACH'].astype('float32')
    df['ORDERLINENUMBER'] = df['ORDERLINENUMBER'].astype('int8')
    df['MSRP'] = df['MSRP'].astype('int16')
    df['WORTH'] = df['WORTH'].astype('int32')

    # Converting categorical columns
    df['PRODUCTLINE'] = df['PRODUCTLINE'].astype('category')
    df['COUNTRY'] = df['COUNTRY'].astype('category')
    df['DEALSIZE'] = df['DEALSIZE'].astype('category')
    
    print("\nMemory usage after optimization:")
    df.info(memory_usage='deep')

# Main menu function
def menu():
    while True:
        print("\nMenu:")
        print("1. Describe SALES")
        print("2. Describe QUANTITYORDERED")
        print("3. Describe PRICEEACH")
        print("4. Describe ORDERLINENUMBER")
        print("5. Describe YEAR_ID")
        print("6. Describe MONTH_ID")
        print("7. Describe PRODUCTLINE")
        print("8. Describe MSRP")
        print("9. Describe PRODUCTCODE")
        print("10. Describe CUSTOMERNAME")
        print("11. Describe PHONE")
        print("12. Describe ADDRESSLINE1")
        print("13. Describe ADDRESSLINE2")
        print("14. Describe CITY")
        print("15. Describe STATE")
        print("16. Describe POSTALCODE")
        print("17. Describe COUNTRY")
        print("18. Describe TERRITORY")
        print("19. Describe CONTACTLASTNAME")
        print("20. Describe CONTACTFIRSTNAME")
        print("21. Describe DEALSIZE")
        print("22. Describe FLAG")
        print("23. Describe WORTH")
        print("24. Describe QTR_ID")
        print("25. Describe YEAR_BIRTH")
        print("26. Add a new record")
        print("27. Optimize memory usage")
        print("28. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            describe_column('SALES')
        elif choice == '2':
            describe_column('QUANTITYORDERED')
        elif choice == '3':
            describe_column('PRICEEACH')
        elif choice == '4':
            describe_column('ORDERLINENUMBER')
        elif choice == '5':
            describe_column('YEAR_ID')
        elif choice == '6':
            describe_column('MONTH_ID')
        elif choice == '7':
            describe_column('PRODUCTLINE')
        elif choice == '8':
            describe_column('MSRP')
        elif choice == '9':
            describe_column('PRODUCTCODE')
        elif choice == '10':
            describe_column('CUSTOMERNAME')
        elif choice == '11':
            describe_column('PHONE')
        elif choice == '12':
            describe_column('ADDRESSLINE1')
        elif choice == '13':
            describe_column('ADDRESSLINE2')
        elif choice == '14':
            describe_column('CITY')
        elif choice == '15':
            describe_column('STATE')
        elif choice == '16':
            describe_column('POSTALCODE')
        elif choice == '17':
            describe_column('COUNTRY')
        elif choice == '18':
            describe_column('TERRITORY')
        elif choice == '19':
            describe_column('CONTACTLASTNAME')
        elif choice == '20':
            describe_column('CONTACTFIRSTNAME')
        elif choice == '21':
            describe_column('DEALSIZE')
        elif choice == '22':
            describe_column('FLAG')
        elif choice == '23':
            describe_column('WORTH')
        elif choice == '24':
            describe_column('QTR_ID')
        elif choice == '25':
            describe_column('YEAR_BIRTH')
        elif choice == '26':
            add_record()
        elif choice == '27':
            optimize_memory()
        elif choice == '28':
            print("Exiting...")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == "__main__":
    menu()
```
