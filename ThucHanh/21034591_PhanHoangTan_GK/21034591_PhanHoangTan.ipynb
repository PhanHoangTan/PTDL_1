PHẦN 3: THỐNG KÊ SUY DIỄN VÀ HỒI QUY TUYẾN TÍNH (5 ĐIỂM – CLO3)

```python
# 1.	Thực hiện kiểm định trung bình của 2 biến số (định lượng) bằng phương pháp Z-Test bằng một giá trị nào đó với mức sai lầm là 10%. Cho nhận xét
from scipy import stats
# Giả sử chúng ta có dữ liệu của hai biến Grade1 và Grade2
# Tạo dữ liệu mẫu cho Grade1 và Grade2
# Lọc dữ liệu cho Grade1 và Grade2 theo điều kiện
grade1 = df[(df['Grade'] >= 6) & (df['Grade'] <= 11)]['Grade']
grade2 = df[(df['Grade'] > 11) & (df['Grade'] <= 16)]['Grade']

# Bước 1: Tính toán các tham số
mean1 = np.mean(grade1)
mean2 = np.mean(grade2)
std1 = np.std(grade1, ddof=1)  # Độ lệch chuẩn mẫu
std2 = np.std(grade2, ddof=1)
n1 = len(grade1)
n2 = len(grade2)

# Bước 2: Tính giá trị Z
Z = (mean1 - mean2) / np.sqrt((std1**2 / n1) + (std2**2 / n2))

# Tính giá trị p-value cho kiểm định hai đuôi
p_value = 2 * (1 - stats.norm.cdf(abs(Z)))

# So sánh với mức sai lầm 10%
alpha = 0.10
print(f"Z-Statistic: {Z:.4f}")
print(f"P-value: {p_value:.4f}")

# Bước 3: Kết luận
if p_value < alpha:
    print("Bác bỏ giả thuyết không (H0). Trung bình của hai biến khác nhau.")
    print("Nhận xét: Có bằng chứng để kết luận rằng trung bình của hai biến khác nhau với mức ý nghĩa 10%.")
else:
    print("Không bác bỏ giả thuyết không (H0). Trung bình của hai biến có thể bằng nhau.")
    print("Nhận xét: Không có đủ bằng chứng để kết luận rằng trung bình của hai biến khác nhau với mức ý nghĩa 10%.")
```

```python
# 1.	Thực hiện kiểm định trung bình của 1 biến số (định lượng) bằng phương pháp Z-Test bằng một giá trị nào đó với mức sai lầm là 5%. Cho nhận xét
# 1.	Thực hiện kiểm định trung bình của 1 biến số (định lượng) bằng phương pháp Z-Test bằng một giá trị nào đó với mức sai lầm là 5%. Cho nhận xét
from statsmodels.stats.weightstats import ztest
import scipy.stats as stats
# Xác định giá trị trung bình giả thuyết
mu_0 = 7  # Giá trị trung bình giả thuyết (có thể thay đổi)

# Bước 1: Tính toán các thông số cần thiết
sample_mean = df['Grade'].mean()  # Trung bình mẫu
sample_std = df['Grade'].std()  # Độ lệch chuẩn mẫu
n = len(df['Grade'])  # Kích thước mẫu

# Bước 2: Tính toán giá trị Z
Z = (sample_mean - mu_0) / (sample_std / np.sqrt(n))

# Tính giá trị p-value cho kiểm định hai đuôi
p_value = 2 * (1 - stats.norm.cdf(abs(Z)))

# Bước 3: Tìm giá trị Z-crit cho mức sai lầm 5%
alpha = 0.05
z_crit = stats.norm.ppf(1 - alpha/2)  # Z-crit cho kiểm định hai đuôi

# Bước 4: Kết luận
print(f"Giá trị Z tính được: {Z:.2f}")
print(f"Giá trị p-value: {p_value:.4f}")
print(f"Giá trị Z-crit: {z_crit:.2f}")

if abs(Z) > z_crit:
    print("Bác bỏ giả thuyết không (H0). Trung bình của biến số khác giá trị giả thuyết.")
    print("Nhận xét: Có bằng chứng để kết luận rằng trung bình của biến số 'Grade' khác với giá trị giả thuyết 7 với mức ý nghĩa 5%.")
else:
    print("Không bác bỏ giả thuyết không (H0). Trung bình của biến số không khác giá trị giả thuyết.")
    print("Nhận xét: Không có đủ bằng chứng để kết luận rằng trung bình của biến số 'Grade' khác với giá trị giả thuyết 7 với mức ý nghĩa 5%.")
```

```python
# 2.Thực hiện kiểm định trung bình của 2 biến số (định lượng) có bằng nhau không bằng phương pháp T-Test với mức sai lầm là 10%. Cho nhận xét
import scipy.stats as stats

# Tách dữ liệu thành hai nhóm: "High" và "Low"
group_high = df[df['Grade Group'] == 'High']['Grade']
group_low = df[df['Grade Group'] == 'Low']['Grade']

# Bước 1: Thực hiện kiểm định T-Test
t_statistic, p_value = stats.ttest_ind(group_high, group_low)

# Bước 2: Xác định mức sai lầm
alpha = 0.10

# Bước 3: Kết luận
print(f"Giá trị T-Statistic: {t_statistic:.4f}")
print(f"Giá trị p-value: {p_value}")

if p_value < alpha:
    print("Bác bỏ giả thuyết không (H0). Trung bình của hai biến không bằng nhau.")
    print("Nhận xét: Có bằng chứng để kết luận rằng trung bình của hai nhóm 'High' và 'Low' khác nhau với mức ý nghĩa 10%.")
else:
    print("Không bác bỏ giả thuyết không (H0). Trung bình của hai biến có thể bằng nhau.")
    print("Nhận xét: Không có đủ bằng chứng để kết luận rằng trung bình của hai nhóm 'High' và 'Low' khác nhau với mức ý nghĩa 10%.")
```

```python
# 4.	Thực hiện kiểm định trung bình của 1 biến số (định lượng) có bằng nhau không bằng phương pháp T-Test với mức sai lầm là 5%. Cho nhận xét
import scipy.stats as stats

# Giả sử biến định lượng cần kiểm định là 'Grade'
grade = df['Grade']  # Thay bằng tên biến cụ thể trong dataframe df

# Xác định giá trị trung bình giả thuyết
mu_0 = 7  # Giá trị trung bình giả thuyết

# Bước 2: Thực hiện kiểm định T-Test
t_statistic, p_value = stats.ttest_1samp(grade, mu_0)

# So sánh với mức sai lầm 5%
alpha = 0.05
print(f"T-Statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Bước 3: Kết luận
if p_value < alpha:
    print("Bác bỏ giả thuyết không (H0). Trung bình của biến số khác giá trị giả thuyết.")
    print("Nhận xét: Có bằng chứng để kết luận rằng trung bình của biến số 'Grade' khác với giá trị giả thuyết 7 với mức ý nghĩa 5%.")
else:
    print("Không bác bỏ giả thuyết không (H0). Trung bình của biến số không khác giá trị giả thuyết.")
    print("Nhận xét: Không có đủ bằng chứng để kết luận rằng trung bình của biến số 'Grade' khác với giá trị giả thuyết 7 với mức ý nghĩa 5%.")
```

```python
# 3.	Thực hiện kiểm tra 2 biến định lượng có tương quan với nhau không bằng phương pháp Chi-Square với mức sai lầm là 5%? Cho nhận xét.
# Giả định: Tạo một biến định lượng khác nếu chưa có trong df
np.random.seed(0)
df['Extra Variable'] = np.random.rand(len(df)) * 10  # Tạo biến ngẫu nhiên

# Bước 1: Chia "Grade" thành các nhóm
# Chia thành 3 nhóm: Low, Medium, High
df['Grade Binned'] = pd.cut(df['Grade'], bins=[0, 6, 11, 16], labels=['Low', 'Medium', 'High'])

# Chia "Extra Variable" thành 2 nhóm
df['Extra Variable Binned'] = pd.cut(df['Extra Variable'], bins=[0, 5, 10], labels=['Low', 'High'])

# Bước 2: Tạo ma trận phân tích
contingency_table = pd.crosstab(df['Grade Binned'], df['Extra Variable Binned'])

# Bước 3: Thực hiện kiểm định Chi-Square
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)

# Bước 4: Kết luận
alpha = 0.05

print("Ma trận phân tích:")
print(contingency_table)
print(f"\nGiá trị Chi-Square: {chi2_stat:.4f}")
print(f"Giá trị p-value: {p_value:.4f}")

if p_value < alpha:
    print("Bác bỏ giả thuyết không (H0). Hai biến có tương quan với nhau.")
    print("Nhận xét: Có bằng chứng để kết luận rằng hai biến 'Grade' và 'Extra Variable' có tương quan với nhau với mức ý nghĩa 5%.")
else:
    print("Không bác bỏ giả thuyết không (H0). Hai biến không có tương quan.")
    print("Nhận xét: Không có đủ bằng chứng để kết luận rằng hai biến 'Grade' và 'Extra Variable' có tương quan với nhau với mức ý nghĩa 5%.")
```

```python
# 4.	Thực hiện kiểm định ANOVA 2 chiều. Giải thích kết quả.
# Giả định: Tạo một biến định lượng khác nếu chưa có trong df
np.random.seed(0)
df['Extra Variable'] = np.random.rand(len(df)) * 10  # Tạo biến ngẫu nhiên

# Bước 1: Tạo các nhóm cho biến "Grade" và các yếu tố phân loại
df['Grade Binned'] = pd.cut(df['Grade'], bins=[0, 6, 11, 16], labels=['Low', 'Medium', 'High'])
df['Extra Variable Binned'] = pd.cut(df['Extra Variable'], bins=[0, 5, 10], labels=['Low', 'High'])

# Tạo DataFrame cho ANOVA
anova_data = df[['Grade', 'Grade Binned', 'Extra Variable Binned']]

# Bước 2: Tạo danh sách cho từng nhóm trong ANOVA
groups = anova_data.groupby(['Grade Binned', 'Extra Variable Binned'])['Grade'].apply(list)

# Thực hiện ANOVA hai chiều
anova_result = stats.f_oneway(*groups)

# Bước 3: Kết luận
print(f"Giá trị F: {anova_result.statistic:.4f}")
print(f"Giá trị p-value: {anova_result.pvalue:.4f}")

alpha = 0.05
if anova_result.pvalue < alpha:
    print("Bác bỏ giả thuyết không (H0). Ít nhất một nhóm có sự khác biệt có ý nghĩa.")
    print("Nhận xét: Có bằng chứng để kết luận rằng ít nhất một nhóm có sự khác biệt có ý nghĩa với mức ý nghĩa 5%.")
else:
    print("Không bác bỏ giả thuyết không (H0). Không có sự khác biệt có ý nghĩa giữa các nhóm.")
    print("Nhận xét: Không có đủ bằng chứng để kết luận rằng có sự khác biệt có ý nghĩa giữa các nhóm với mức ý nghĩa 5%.")
```

```python
# 5.	Đánh giá mức độ tác động giữa các yếu tố đến 1 đối tượng (1 biến độc lập và 1 biến phụ thuộc) bằng phân tích hồi quy tuyến tính theo phương thức OLS. Giải thích kết quả
import statsmodels.api as sm
# Tạo một biến độc lập và biến phụ thuộc
X = df['Extra Variable']  # Biến độc lập
Y = df['Grade']  # Biến phụ thuộc

# Bước 1: Thêm hằng số vào biến độc lập
X = sm.add_constant(X)

# Bước 2: Thực hiện hồi quy tuyến tính
model = sm.OLS(Y, X).fit()

# Bước 3: Tóm tắt kết quả hồi quy
summary = model.summary()
print(summary)

# Lấy hệ số hồi quy
coefficients = model.params

# Giải thích kết quả
p_values = model.pvalues

print(f'Coefficients:\n{coefficients}')
print(f'P-values:\n{p_values}')

alpha = 0.05

for param, p_value in p_values.items():
    if p_value < alpha:
        print(f'Từ chối giả thuyết H0 cho {param}')
        print(f'Nhận xét: {param} có tác động đáng kể đến biến phụ thuộc với mức sai lầm 5%.')
    else:
        print(f'Chấp nhận giả thuyết H0 cho {param}')
        print(f'Nhận xét: {param} không có tác động đáng kể đến biến phụ thuộc với mức sai lầm 5%.')

# Viết phương trình hồi quy tuyến tính
intercept = coefficients['const']
slope = coefficients['Extra Variable']
print(f'Phương trình hồi quy tuyến tính: Grade = {intercept:.2f} + {slope:.4f} * Extra Variable')

# Vẽ biểu đồ dữ liệu
plt.scatter(df['Extra Variable'], df['Grade'], color='blue', label='Data Points')

# Vẽ đường hồi quy
regression_line = intercept + slope * df['Extra Variable']
plt.plot(df['Extra Variable'], regression_line, color='red', label='Regression Line')

# Thêm nhãn và tiêu đề
plt.xlabel('Extra Variable')
plt.ylabel('Grade')
plt.title('Hồi Quy Tuyến Tính: Grade vs. Extra Variable')
plt.legend()

# Hiển thị biểu đồ
plt.show()

```
